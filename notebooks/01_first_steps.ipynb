# First steps: run a model, change a parameter, find equilibria

The Morris-Lecar model is a two-variable model of a neuron's membrane: a
voltage `V` and a recovery variable `W`, driven by an applied current
`iapp`. It ships with XPP Lab. In this notebook you load it, run it, plot
the run, change `iapp` and find the equilibria.

Run the cells in order (Shift+Enter).

```python
import matplotlib.pyplot as plt
import xpplab as xs

# xs.example finds a model that ships with the package
m = xs.load(xs.example("lecar.ode"))
m
```

## Run it

`m.run()` integrates with the model's own settings (the Runge-Kutta method,
from the `.ode` file) and returns a trajectory. `total` is the length of
the run in time units.

```python
m.parameters["iapp"]
```

```python
sol = m.run(total=100)
sol.columns
```

```python
sol.plot("V", "W");
```

With `iapp = 0.05` the cell is excitable: it starts away from its resting
state, makes one spike and settles. The phase plane shows the same run in
the (`V`, `W`) plane with the nullclines and the vector field.

```python
m.phase_plane();
```

## Change a parameter

`m.parameters` is a dict of the model's parameters. Raise the applied
current and run again.

```python
m.parameters["iapp"] = 0.1
sol = m.run(total=300)
sol.plot("V");
```

```python
late = sol["V"][sol.t > 200]
late.min(), late.max()
```

Now the cell fires without stopping: after the first 200 time units `V`
still swings between about -0.38 and 0.26. A small change of one
parameter turned a resting cell into an oscillator.

```python
m.phase_plane();
```

## Find the equilibria

`m.find_equilibria()` starts Newton's method from many points and returns
the distinct equilibria, each with its eigenvalues and its stability.

```python
eqs = m.find_equilibria()
eqs
```

```python
for e in eqs:
    print(e.values, e.eigenvalues, "stable" if e.stable else "unstable")
```

There is a single equilibrium and it is unstable, with a pair of complex
eigenvalues whose real part is positive: trajectories spiral away from it
and onto the oscillation. Put `iapp` back to 0.05 and ask again: there
are three equilibria, and one of them is a stable node, the resting state.

```python
m.parameters["iapp"] = 0.05
m.find_equilibria()
```
